## Start with the deep learning model
**Phase 2 trains a neural network: a denoising autoencoder.** Four learned Linear layers compress 29 features into 8 values, then rebuild 29 outputs. Hidden ReLU activations make the mapping nonlinear. Backpropagation and Adam update the network's 1,253 parameters.

**Input during training:** a noisy legitimate transaction. **Target:** the same clean transaction. The model learns feature values, rather than a fraud label.

Read [deep learning in 10 minutes](../output/DEEP_LEARNING_IN_10_MINUTES.html) first. Notebook 03 shows the real network, a shape trace, the training loop and learning curves. Notebook 04 evaluates the trained AE and then adds the hybrid. GMM remains the Phase 1 comparison.

## The whole project in five sentences
1. We learn normal spending from earlier **legitimate transactions only**.
2. A **GMM** asks: how unusual is this transaction under a mixture of normal patterns?
3. An **autoencoder (AE)** asks: how badly do I rebuild this transaction after compressing it?
4. Our **hybrid** combines these different scores after putting them on comparable scales.
5. We compare the methods on the same later mixed transactions, counting fraud caught and false alerts.

## Why an autoencoder?
A fraud classifier needs fraud examples as training targets. Our AE instead learns to copy normal inputs.
A small middle layer forces it to learn useful structure. Some unusual transactions are harder to copy.
**That is a hypothesis, not a guarantee:** fraud can look normal and reconstruct well.

| Method | Learns | Score (larger = more unusual) | Main weakness |
|---|---|---|---|
| GMM | Mixture of Gaussian densities | Negative log density | Shape assumptions; diagonal covariance ignores within-component relationships |
| AE | Nonlinear reconstruction | Mean squared reconstruction error | Can reconstruct fraud well |
| Hybrid | No new neural network | Equal average of normalized GMM and AE scores | May combine correlated errors and worsen ranking |

## Why is this hybrid meaningful?
The density model and reconstruction model ask different questions. We test whether their combined
ranking helps under a real review limit. Side-by-side comparison alone is **not** our hybrid.
The implemented score fusion is the hybrid. We report any loss as well as any gain.

## Experiment rules fixed before running Phase 2
- Same deduplication assumption and chronological 60/20/20 split as Phase 1.
- Same 29 inputs and same legitimate training rows for GMM and AE.
- AE training target = its own clean input. Never supply fraud as a target.
- Main AE: 29-16-8-16-29, ReLU inside, linear output, input noise SD 0.10 during training.
- Adam, learning rate 0.001, batch size 1024, weight decay 0.00001, at most 30 epochs.
- Stop after 5 epochs without a normal-validation MSE improvement of at least 0.0001.
- Main model fixed in advance. Plain AE, bottleneck 12, and a second seed are diagnostic ablations.
- Hybrid: fixed equal weights, normalization learned from normal validation scores. No label-based weight search.
- Validation FPR policies: 0.1%, 0.5%, 1%. Default = 0.5%. Review capacities: 50, 100, 200, 500.
- Report trapezoidal PR AUC **and** average precision (AP), which use different definitions.
- Phase 1 already exposed this test benchmark. Phase 2 does not use test labels to tune its models,
  but this is a reused benchmark, not a newly blind test. Future data would be needed for that.
